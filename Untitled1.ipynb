```python
import glob, numpy as np, pandas as pd
from sklearn.metrics import roc_auc_score

EXCLUDE = {("deeplabv3", 3), ("deeplabv3", 4)}   # invalid runs; use set() after retraining
NAMES  = {"unet": "U-Net", "attunet": "Attention U-Net", "deeplabv3": "DeepLabV3"}
PARAMS = {"unet": 24.44, "attunet": 24.79, "deeplabv3": 26.01}

df = pd.concat([pd.read_csv(f) for f in glob.glob("results/per_image_*_fold*.csv")], ignore_index=True)
df = df[[(m, int(k)) not in EXCLUDE for m, k in zip(df.model, df.fold)]]
df["is_positive"] = df.is_positive.astype(bool)
pos = df[df.is_positive]

rows = [("dice","Dice"),("iou","IoU"),("precision","Precision"),("recall","Recall"),
        ("hd95_mm","HD95 (mm)"),("assd_mm","ASSD (mm)"),("nsd_1mm","NSD @ 1 mm"),("nsd_2mm","NSD @ 2 mm"),
        ("lesion_recall_02","Lesion recall (IoU>=0.2)"),("lesion_precision_02","Lesion precision (IoU>=0.2)"),
        ("lesion_f1_02","Lesion F1 (IoU>=0.2)"),("lesion_recall_05","Lesion recall (IoU>=0.5)")]
rows = [r for r in rows if r[0] in pos.columns]

pf = pos.groupby(["model","fold"])[[r[0] for r in rows]].mean()
pf["missed_pct"] = (pos.dice == 0).groupby([pos.model, pos.fold]).mean() * 100
pf["fppi"] = df.groupby(["model","fold"]).fp_lesions_02.mean()
if "inference_ms" in df: pf["ms"] = df.groupby(["model","fold"]).inference_ms.mean()
pf["auc_roc"] = pd.Series({(m, k): roc_auc_score(g.is_positive, g.max_prob)
                           for (m, k), g in df.groupby(["model","fold"]) if g.is_positive.nunique() == 2})

labels = dict(rows)
labels.update(missed_pct="Lesion images with Dice = 0 (%)", fppi="FP lesions per image (all images)",
              auc_roc="AUC-ROC (image level)", ms="Inference (ms/image)")

out = {}
for m, n in NAMES.items():
    if m not in pf.index.get_level_values(0): continue
    g = pf.loc[m]
    out[n] = {labels[c]: f"{g[c].mean():.3f} ± {g[c].std(ddof=1):.3f}" for c in labels if c in g}
    out[n]["Parameters (M)"] = PARAMS[m]
    out[n]["Folds used"] = len(g)

table = pd.DataFrame(out)
table.to_csv("results/final_table.csv")
try: open("results/final_table.md", "w").write(table.to_markdown())
except Exception: pass
display(table)

# per-fold table too (useful for the report)
pf.round(3).to_csv("results/per_fold_table.csv"); display(pf.round(3))

# back up results so a runtime reset doesn't lose them
!cp -r results /content/drive/MyDrive/INbreast_results
```
